```python
import cv2
import random
import numpy as np
import os
```

```python
input_path = "day_frame"
output_path = "rain_frame"

os.makedirs(output_path, exist_ok=True)
```

```python
def add_rain(image):

    rain = image.copy()

    h, w = rain.shape[:2]

    rain_count = 6000

    for _ in range(rain_count):

        x = np.random.randint(0, w)
        y = np.random.randint(0, h)

        length = np.random.randint(20, 40)

        cv2.line(
            rain,
            (x, y),
            (x + 4, y + length),
            (200, 200, 200),
            1
        )

    # เบลอเส้นฝน
    rain = cv2.GaussianBlur(rain, (3, 3), 0)

    # ทำให้ภาพมืดลง
    rain = cv2.convertScaleAbs(
        rain,
        alpha=0.9,
        beta=-20
    )

    # -----------------------
    # Fog
    # -----------------------

    fog_strength = random.uniform(0.6, 0.8)

    fog = np.random.normal(
        220,
        15,
        rain.shape
    ).astype(np.uint8)

    fog = cv2.GaussianBlur(
        fog,
        (101, 101),
        0
    )

    gradient = np.linspace(
        1.0,
        0.5,
        h
    ).reshape(h, 1, 1)

    fog = (fog * gradient).astype(np.uint8)

    rain = cv2.addWeighted(
        rain,
        1 - fog_strength,
        fog,
        fog_strength,
        0
    )

    return rain
```

```python
img_count = 0

for file_name in os.listdir(input_path):

    if not file_name.lower().endswith((".jpg", ".jpeg", ".png")):
        continue

    image_path = os.path.join(input_path, file_name)

    img = cv2.imread(image_path)

    if img is None:
        print(f"อ่านไม่ได้: {file_name}")
        continue

    rain_img = add_rain(img)

    save_path = os.path.join(
        output_path,
        f"frame4_{img_count:04d}.jpg"
    )

    cv2.imwrite(save_path, rain_img)

    img_count += 1

print(f"Done! Saved {img_count} images.")
```
